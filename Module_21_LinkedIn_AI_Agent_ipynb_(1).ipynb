```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
You are a professional LinkedIn content writer.

Your task is to generate high-quality LinkedIn posts.

Requirements:
- Write in the language requested by the user.
- Write 2 to 4 paragraphs.
- Start with an engaging hook.
- Provide useful and relevant insights.
- Use a professional but natural tone.
- Make the post suitable for LinkedIn.
- End with 3 to 5 relevant hashtags.
- Do not mention that you are an AI.
- Do not use unnecessary headings.
"""
    ),
    (
        "human",
        """
Topic: {topic}

Language: {language}

Generate the LinkedIn post now.
"""
    )
])

print("✅ LinkedIn prompt created successfully!")
```

```python
chain = prompt | llm

print("✅ LangChain chain created successfully!")
```

```python
def generate_linkedin_post(topic, language):

    if not topic.strip():
        return "Please enter a topic."

    if not language.strip():
        return "Please enter a language."

    response = chain.invoke({
        "topic": topic,
        "language": language
    })

    return response.content

print("✅ LinkedIn Agent function created!")
```

```python
import gradio as gr
```

```python
def linkedin_agent(topic, language):
    return generate_linkedin_post(topic, language)
```

```python
with gr.Blocks(
    title="LinkedIn AI Agent"
) as demo:

    gr.Markdown(
        """
        # 🤖 LinkedIn AI Agent

        Generate professional LinkedIn posts using
        **LangChain + Groq**.

        Enter a topic and select your preferred language.
        """
    )

    topic = gr.Textbox(
        label="📌 Post Topic",
        placeholder="e.g. AI in Healthcare",
        lines=2
    )

    language = gr.Dropdown(
        choices=[
            "English",
            "Bengali",
            "Spanish",
            "Hindi"
        ],
        value="English",
        label="🌐 Language"
    )

    generate_button = gr.Button(
        "🚀 Generate LinkedIn Post"
    )

    output = gr.Textbox(
        label="📝 Generated LinkedIn Post",
        lines=15
    )

    generate_button.click(
        fn=linkedin_agent,
        inputs=[topic, language],
        outputs=output
    )

demo.launch()
```
